```python
import pandas as pd
import matplotlib.pyplot as plt

input_csv = pd.read_csv(
r"C:\Users\Kohsu\Desktop\graduation_thesis\progress\data\data.csv"
)

# x軸：四半期を文字列として扱う
x = input_csv.iloc[:, 0].astype(str)

# y軸：R
R = input_csv["R"]/100 + 1

plt.figure(figsize=(8, 4))

plt.plot(
x,
R,
linestyle="solid",
marker="o"
)

plt.ylabel("%")
plt.xticks(
    range(0, len(x), 2),
    x.iloc[::2],
    rotation=90
)
plt.title("R")

plt.grid(True)

plt.tight_layout()
plt.savefig("R_data.png", dpi=300, bbox_inches="tight")
plt.show()

```
